## Ba quan sát chính (đối chiếu lý thuyết)

1. **Zero & Random (naive) không bao giờ vượt qua mức ngẫu nhiên (~10%)** ở cả 4 activation — khớp Phase 2 mục 5.1 (mất symmetry breaking với Zero) và mục 5.2/6 (Random không scale theo fan-in nên vẫn vanishing dù đã phá đối xứng).
2. **Sigmoid thất bại với MỌI initialization** ở độ sâu 6 lớp trong 15 epoch/SGD thuần — kể cả He/Xavier vốn "đúng lý thuyết". Nguyên nhân: $\sigma'(z)\le 0.25$ luôn luôn, nên hệ số suy giảm tối đa mỗi lớp là 0.25 CHO DÙ weight scale đã đúng — sau 6 lớp hệ số trần đã là $0.25^6\approx 2.4\times10^{-4}$ (Phase 2 mục 6). Đây là bằng chứng thực nghiệm cho giới hạn của initialization: nó giảm nhẹ nhưng không xoá được vanishing gradient gây ra bởi chính activation bão hoà.
3. **LeCun/Xavier/He đều hoạt động tốt với Tanh/ReLU/Leaky ReLU** (~64–70%), sai khác giữa 3 scheme này nhỏ hơn nhiều so với sai khác Zero/Random vs. có-scale — khớp việc cả ba đều bảo toàn variance tín hiệu ở bậc-độ-lớn đúng, chỉ khác hệ số nhân 1 so với 2.
